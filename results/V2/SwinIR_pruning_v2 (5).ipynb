```python
import os
from PIL import Image
import torch.nn.functional as F
import torchvision.transforms as transforms

to_tensor = transforms.ToTensor()
to_pil = transforms.ToPILImage()

def create_lr_dataset(root_path, scale):
    hr_dir = os.path.join(root_path, "HR")
    lr_dir = os.path.join(root_path, f"x{scale}")
    os.makedirs(lr_dir, exist_ok=True)

    for filename in os.listdir(hr_dir):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):
            # Load HR image
            img_path = os.path.join(hr_dir, filename)
            img = Image.open(img_path).convert("RGB")

            # Convert to tensor
            tensor = to_tensor(img).unsqueeze(0)  # shape: (1, C, H, W)
            _, _, h, w = tensor.shape

            # Downsample with bicubic interpolation
            down = F.interpolate(
                tensor,
                size=(h // scale, w // scale),
                mode="bicubic",
                align_corners=False
            )

            # Convert back to image
            img_down = to_pil(down.squeeze(0))

            # Save in LR folder
            save_path = os.path.join(lr_dir, filename)
            img_down.save(save_path)

    print(f"✅ Created LR dataset at {lr_dir}")


parent_dir = "/content/TEST_SETS"
scale = 2  # or 4

for subdir in os.listdir(parent_dir):
    full_path = os.path.join(parent_dir, subdir)
    if os.path.isdir(full_path):  # only process directories
        create_lr_dataset(full_path, scale=scale)
```
