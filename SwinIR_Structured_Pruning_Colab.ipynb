## 🖼️ Step 8: Visualize Results (Optional)

```python
# Visualize some results if images are available
import matplotlib.pyplot as plt
from PIL import Image
import glob

def show_comparison_images(image_dirs, max_images=3):
    """Show comparison between baseline and pruned results."""
    
    baseline_dirs = [d for d in image_dirs if 'baseline' in d]
    pruned_dirs = [d for d in image_dirs if 'pruned' in d and 'iter' in d]
    
    if not baseline_dirs or not pruned_dirs:
        print("❌ No baseline or pruned image directories found.")
        return
    
    # Get latest pruned iteration
    latest_pruned = sorted(pruned_dirs)[-1]
    
    baseline_path = os.path.join(images_dir, baseline_dirs[0])
    pruned_path = os.path.join(images_dir, latest_pruned)
    
    # Find corresponding images
    baseline_images = glob.glob(os.path.join(baseline_path, "*.png"))[:max_images]
    
    if not baseline_images:
        print("❌ No baseline images found.")
        return
    
    fig, axes = plt.subplots(len(baseline_images), 3, figsize=(15, 5*len(baseline_images)))
    if len(baseline_images) == 1:
        axes = axes.reshape(1, -1)
    
    for i, baseline_img_path in enumerate(baseline_images):
        img_name = os.path.basename(baseline_img_path)
        pruned_img_path = os.path.join(pruned_path, img_name)
        
        # Find original low-res image
        lr_img_path = None
        for lr_path in glob.glob("testsets/set5/LR_bicubic/X2/*.png"):
            if os.path.basename(lr_path).split('.')[0] in img_name:
                lr_img_path = lr_path
                break
        
        # Load images
        try:
            if lr_img_path and os.path.exists(lr_img_path):
                lr_img = Image.open(lr_img_path)
                axes[i, 0].imshow(lr_img)
                axes[i, 0].set_title(f"Low Resolution\n{lr_img.size}")
                axes[i, 0].axis('off')
            else:
                axes[i, 0].text(0.5, 0.5, "LR Image\nNot Found", ha='center', va='center')
                axes[i, 0].axis('off')
            
            if os.path.exists(baseline_img_path):
                baseline_img = Image.open(baseline_img_path)
                axes[i, 1].imshow(baseline_img)
                axes[i, 1].set_title(f"Baseline SR\n{baseline_img.size}")
                axes[i, 1].axis('off')
            
            if os.path.exists(pruned_img_path):
                pruned_img = Image.open(pruned_img_path)
                axes[i, 2].imshow(pruned_img)
                axes[i, 2].set_title(f"Pruned SR\n{pruned_img.size}")
                axes[i, 2].axis('off')
            else:
                axes[i, 2].text(0.5, 0.5, "Pruned Image\nNot Found", ha='center', va='center')
                axes[i, 2].axis('off')
        
        except Exception as e:
            print(f"Error loading image {img_name}: {e}")
    
    plt.tight_layout()
    plt.show()

# Try to show comparison images
if os.path.exists(images_dir):
    image_dirs = [d for d in os.listdir(images_dir) if os.path.isdir(os.path.join(images_dir, d))]
    if image_dirs:
        print("🖼️ Showing comparison images...")
        show_comparison_images(image_dirs, max_images=2)
    else:
        print("❌ No image directories found.")
else:
    print("❌ Images directory not found.")
```

## 📥 Step 9: Download Results

```python
# Prepare results for download
import zipfile
from google.colab import files

def create_results_zip():
    """Create a ZIP file with all results."""
    
    zip_filename = "swinir_structured_pruning_results.zip"
    
    with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
        # Add model files
        models_dir = "superresolution/swinir_sr_structured_pruning/models"
        if os.path.exists(models_dir):
            for root, dirs, files in os.walk(models_dir):
                for file in files:
                    file_path = os.path.join(root, file)
                    arc_path = os.path.relpath(file_path, "superresolution")
                    zipf.write(file_path, arc_path)
        
        # Add a sample of result images (to keep zip size manageable)
        images_dir = "superresolution/swinir_sr_structured_pruning/images"
        if os.path.exists(images_dir):
            # Only include a few sample images
            sample_count = 0
            max_samples = 10
            
            for root, dirs, files in os.walk(images_dir):
                for file in files[:2]:  # Max 2 files per directory
                    if sample_count >= max_samples:
                        break
                    if file.endswith('.png'):
                        file_path = os.path.join(root, file)
                        arc_path = os.path.relpath(file_path, "superresolution")
                        zipf.write(file_path, arc_path)
                        sample_count += 1
                if sample_count >= max_samples:
                    break
        
        # Add log files
        log_dir = "superresolution/swinir_sr_structured_pruning"
        if os.path.exists(log_dir):
            for file in os.listdir(log_dir):
                if file.endswith('.log') or file.endswith('.txt'):
                    file_path = os.path.join(log_dir, file)
                    arc_path = os.path.relpath(file_path, "superresolution")
                    zipf.write(file_path, arc_path)
    
    return zip_filename

# Create and download results
try:
    print("📦 Creating results package...")
    zip_filename = create_results_zip()
    
    if os.path.exists(zip_filename):
        zip_size = os.path.getsize(zip_filename) / (1024 * 1024)  # Size in MB
        print(f"✅ Results package created: {zip_filename} ({zip_size:.1f} MB)")
        
        print("\n📥 Downloading results...")
        files.download(zip_filename)
        
        print("\n📋 Package Contents:")
        print("   - Pruned model (.pth files)")
        print("   - Results summary (pruning_results.txt)")
        print("   - Sample output images")
        print("   - Training logs")
        
    else:
        print("❌ Failed to create results package.")
        
except Exception as e:
    print(f"❌ Error creating results package: {e}")
    print("You can manually download individual files from the Files panel.")
```
